```python
import numpy as np
import pandas as pd
```

```python
upwork2=pd.read_csv('upwork2.csv')
```

```python
upwork2.info()
```

```python
upwork2.head(10)
```

```python
keyword_cols=['highlight-color 2','highlight-color 3','air3-token','highlight-color 4','air3-token 2', 'air3-token 4']
```

```python
all_keywords = (
    upwork2[keyword_cols]
    .stack()
    .dropna()
    .str.lower()
    .str.strip()
)
```

```python
all_keywords.nunique()
```

```python
keyword_frequency = all_keywords.value_counts()
```

```python
keyword_frequency.head(15)
```

```python
top15 = keyword_frequency.head(15).reset_index()
top15.columns = ['Keyword', 'Frequency']
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.bar(top15['Keyword'], top15['Frequency'])
plt.xticks(rotation=45, ha='right')
plt.title('Top 15 Most Used UI/UX Keywords')
plt.xlabel('Keyword')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.barh(top15['Keyword'], top15['Frequency'])
plt.title('Top 15 UI/UX Keywords by Demand')
plt.xlabel('Number of Job Listings')
plt.ylabel('Keyword')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
top15['Rank'] = range(1, 16)
top15 = top15[['Rank', 'Keyword', 'Frequency']]
top15
```

```python
top15 = top15.head(15)

plt.figure(figsize=(6,6))
plt.pie(
    top15['Frequency'],
    labels=top15['Keyword'],
    autopct='%1.1f%%',
    startangle=140
)
plt.title('Top UI/UX Keyword Share')
plt.show()
```
